```sql
create database nleshin_catalog.ai_layer;
```

```sql
-- Step 1 - Create a function to search objects description

CREATE OR REPLACE FUNCTION nleshin_catalog.ai_layer.search_objects_description(
  user_question STRING COMMENT 'User question for finding relevant objects description.'
)
RETURNS TABLE
COMMENT '''
Searches object descriptions using semantic vector search to retrieve the most relevant documentation for answering the user''s question.
Use this function when the user asks about data objects, including their purpose, structure, columns, data types, constraints, relationships, or other documented characteristics.
The search is performed against chunked object documentation and returns the most relevant documentation chunks for use as context when generating the answer.
'''
RETURN(
  SELECT
    chunk_text_for_returning
  FROM
    vector_search(
      index => 'nleshin_catalog.ai_layer.objects_description_index',
      query_text => user_question,
      num_results => 5
  )
);
```

```sql
select * from nleshin_catalog.ai_layer.search_objects_description("Кто является владельцем таблицы ");
```
