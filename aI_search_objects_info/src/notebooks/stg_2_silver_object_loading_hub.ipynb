```sql
-- Step 1 - Create hub type table in silver layer

-- drop table nleshin_catalog.silver_layer.objects_description_hub purge;

create table if not exists nleshin_catalog.silver_layer.objects_description_hub(
    file_name string not null,
    file_surrogate_key string not null primary key,
    created_datetime string  not null default current_timestamp,
    source_id int
)
CLUSTER BY AUTO
TBLPROPERTIES('delta.feature.allowColumnDefaults' = 'supported');
```

```sql
-- Step 2 - Load information about key to the hub table

merge into nleshin_catalog.silver_layer.objects_description_hub trg
using (
    select file_name 
    from nleshin_catalog.stg_layer.objects_description
) src
on src.file_name = trg.file_name
when not matched then insert (
  file_name,
  file_surrogate_key
) values (
  src.file_name,
  md5(src.file_name)
);
```
