```sql
-- Use nessessary catalog
use catalog nleshin_catalog;
```

```sql
-- Create new schema
create schema if not exists raw_layer;
```

```sql
-- Create new volume
CREATE VOLUME if not exists nleshin_catalog.raw_layer.orders;
```

```sql
-- Can see content of our files in volume
-- SELECT * FROM read_files('/Volumes/nleshin_catalog/raw_layer/orders');
SELECT * FROM read_files(
    '/Volumes/nleshin_catalog/raw_layer/orders',
    format => 'json',
    schema => 'order_id BIGINT, customer_id BIGINT, amount BIGINT, status STRING'
);
```

```sql
-- Create simple table
CREATE OR REPLACE TABLE nleshin_catalog.bronze_layer.orders (
    order_id BIGINT, 
    customer_id BIGINT, 
    amount BIGINT, 
    status STRING
);
```

```sql
select * from nleshin_catalog.bronze_layer.orders;
```

```sql
-- Load data from files in folder to the table
COPY INTO nleshin_catalog.bronze_layer.orders
FROM '/Volumes/nleshin_catalog/raw_layer/orders'
FILEFORMAT = JSON
```
